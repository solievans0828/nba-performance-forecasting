# 01 — Data Pull (nba_api)

This notebook pulls NBA player game logs using `nba_api` and saves them to `../data/raw/`.

**Why this matters for performance analytics:** building your own dataset pipeline (instead of downloading a CSV) is a real-world analytics workflow.


## 1) Install + Imports

Run the install cell if you're in a fresh environment.


```python
# If needed (e.g., Colab / fresh venv):
# !pip install -r ../requirements.txt

import os
from pathlib import Path
import pandas as pd
```

## 2) Pull game logs for a small starter list

This project uses a small default list to keep things fast and reduce rate-limit issues.

If you want to change the player list later, you can edit the script in `src/pull_player_gamelogs.py`.


```python
import subprocess
import sys

# Pull 2024-25 season logs (change season if needed)
subprocess.check_call([sys.executable, '-m', 'src.pull_player_gamelogs', '--season', '2024-25'])
```

## 3) Load the pulled CSV

The output file is saved to `data/raw/player_gamelogs_2024-25.csv`.


```python
raw_path = Path('..') / 'data' / 'raw' / 'player_gamelogs_2024-25.csv'
df = pd.read_csv(raw_path)
df[['PLAYER_NAME','GAME_DATE','MATCHUP','WL','MIN','PTS','AST','REB']].head(10)
```

## 4) Quick sanity checks

Basic checks to confirm:
- multiple players are present
- date ranges look correct
- no obvious duplicates


```python
print('Rows:', len(df))
print('Players:', df['PLAYER_NAME'].nunique())
print('Date min/max:', df['GAME_DATE'].min(), df['GAME_DATE'].max())

# Example: games per player
df.groupby('PLAYER_NAME').size().sort_values(ascending=False)
```
